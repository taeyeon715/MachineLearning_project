```python
import pandas as pd

# CSV 파일 불러오기
data = pd.read_csv('기계학습_dog.csv')

# 색상 값에 대한 매핑
color_mapping = {
    1: 'black',
    2: 'brown',
    3: 'golden',
    4: 'yellow',
    5: 'cream',
    6: 'gray',
    7: 'white'
}

# Color1, Color2, Color3에 대한 one-hot 인코딩
for color_col in ['Color1', 'Color2', 'Color3']:
    data[color_col] = data[color_col].map(color_mapping)
    data_encoded = pd.get_dummies(data[color_col], prefix=f'Color_{color_col}')

    # 원래 데이터프레임에 인코딩된 컬럼을 추가
    data = pd.concat([data, data_encoded], axis=1)

# One-Hot인코딩으로 안 된 새로운 열 추가 및 모든 값 False로 설정
data['Color_Color2_black'] = False
data['Color_Color3_black'] = False
data['Color_Color3_brown'] = False

# 기존의 컬러 열 삭제
data.drop(['Color1', 'Color2', 'Color3'], axis=1, inplace=True)

# 각 색상에 대해 Color_back 열 추가 및 값을 설정
for color in color_mapping.values():
    # Color1_{color}, Color2_{color}, Color3_{color} 3열의 값 중에 하나라도 TRUE가 있으면 TRUE로, 모두 FALSE인 경우에만 FALSE로 설정
    data[f'Color_{color}'] = data[[f'Color_Color1_{color}', f'Color_Color2_{color}', f'Color_Color3_{color}']].any(axis=1)
    
    # 기존의 Color1_{color}, Color2_{color}, Color3_{color} 열 삭제
    data.drop([f'Color_Color1_{color}', f'Color_Color2_{color}', f'Color_Color3_{color}'], axis=1, inplace=True)
    
# 추가된 열에 대한 처리
for color in color_mapping.values():
    # Color_{color} 열의 값을 TRUE이면 1로, FALSE이면 0으로 변경
    data[f'Color_{color}'] = data[f'Color_{color}'].astype(int)

data.to_csv('기계학습_color_onehot.csv', index=False, encoding='utf-8-sig')
```

```python
#catBoost 최적의 하이퍼파라미터 찾는 코드

from catboost import CatBoostClassifier, Pool
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.model_selection import GridSearchCV

# CSV 파일 불러오기
data = pd.read_csv('기계학습_dog.csv')

# 'PetID' 열 제거
data = data.drop('PetID', axis=1)
data = data.drop('RescuerID', axis=1)

# 특성 변수 선택
features = data[['Age', 'Breed1', 'Breed2', 'Color1', 'Color2', 'Color3', 
                'MaturitySize', 'FurLength', 'Vaccinated', 'Dewormed', 'Sterilized', 
                'Health', 'Fee', 'State', 'VideoAmt', 'PhotoAmt', 'DesSem']]

# 타겟 변수 추출
target = data['AdoptionSpeed']

# 특성과 타겟 변수 분리
X_train, X_test, y_train, y_test = train_test_split(features, target, test_size=0.2, random_state=42)

# 카테고리형 변수 목록
cat_features = ['Breed1', 'Breed2', 'Color1', 'Color2', 'Color3', 'MaturitySize', 'FurLength', 'Vaccinated', 'Dewormed', 'Sterilized', 'Health', 'State']

# 데이터를 Pool 객체로 변환
train_data = Pool(data=X_train, label=y_train, cat_features=cat_features)
test_data = Pool(data=X_test, label=y_test, cat_features=cat_features)

# 하이퍼파라미터 그리드 설정
param_grid = {
    'iterations': [100, 200, 300, 400],  # 부스팅 라운드 수
    'depth': [4, 6, 8, 10],             # 트리 깊이
    'learning_rate': [0.3 ,0.2 ,0.1, 0.01, 0.001]  # 학습률
}

# CatBoost 모델 생성
model = CatBoostClassifier()

# GridSearchCV를 사용하여 최적의 하이퍼파라미터 탐색
grid_search = GridSearchCV(model, param_grid, cv=3)
grid_search.fit(X_train, y_train)

# 최적의 모델 및 하이퍼파라미터 출력
print("최적의 파라미터:", grid_search.best_params_)
print("최적의 모델:", grid_search.best_estimator_)

# 최적의 모델로 예측 수행
predictions = grid_search.predict(X_test)

# 모델의 성능 평가
accuracy = accuracy_score(y_test, predictions)
print("정확도:", accuracy)
```
